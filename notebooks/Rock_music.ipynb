- FeS2, pyrite (soft): mp-1522
- Diamond (hard, high freq): mp-66
- Barium zirconate (soft, low freq): mp-3834

```python
import mp_api
from mp_api.client import MPRester
import ase
from ase.visualize import view
import os
from dotenv import load_dotenv
load_dotenv () # use python-dotenv library for storing secrets in a .env file in project route (or at another path that is specified here)


with MPRester(os.getenv('MP_API_KEY')) as mpr:
    structure = mpr.get_structure_by_material_id("mp-3953")  # Example: Silicon

# Convert Pymatgen Structure to ASE Atoms Object
ase_structure = structure.to_ase_atoms()
view(ase_structure*[3,3,3], viewer='x3d')
```
